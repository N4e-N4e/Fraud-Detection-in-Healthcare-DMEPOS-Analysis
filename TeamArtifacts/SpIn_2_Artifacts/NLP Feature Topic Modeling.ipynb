```python
import pandas as pd
import re
from sklearn.feature_extraction.text import CountVectorizer
import spacy
```

```python
products = pd.read_csv("/dsa/groups/casestudycf25/team02/general_payments_products_clean.csv")
```

```python
products.head(2)
```

```python
text_cols = ["product_name", "product_category"]

def combine_text_cols(row, cols):
    parts = [str(row[c]) for c in cols if pd.notnull(row[c])]
    return " ".join(parts).strip()

products["text_raw"] = products.apply(lambda r: combine_text_cols(r, text_cols), axis=1)

# Drop rows with no text
products = products[products["text_raw"].str.strip().astype(bool)].copy()
```

```python
from spacy.lang.en.stop_words import STOP_WORDS

# Start with spaCy's general English stop words
spacy_stops = set(STOP_WORDS)
```

```python
# text cleaning

def clean_text(s: str) -> str:
    s = s.lower()
    # remove non-letters
    s = re.sub(r"[^a-z0-9\s\-]+", " ", s)
    # collapse whitespace
    s = re.sub(r"\s+", " ", s).strip()
    return s

products["text_clean"] = products["text_raw"].astype(str).map(clean_text)

docs = products["text_clean"].tolist()
```
