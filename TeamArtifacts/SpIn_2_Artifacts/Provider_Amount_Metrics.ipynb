## Provider_Amount_Metrics
Purpose of this notebook is to derive a dataset that contains NPI-level aggregated statistics (mean, sum, median, standard deviation, min, and max) for all amount-related metrics, along with an exclusion flag indicating which NPIs are excluded.

End Result:

Saved the cleaned dataset to a shared team directory: /dsa/groups/casestudycf25/team02/DMEPOS_Amount_stats.csv

```python
# Loading the required Libraries and Dataset
from pathlib import Path
import pandas as pd
df=pd.read_csv("/dsa/groups/casestudycf25/team02/DMEPOS_RefPro_Ser_nar_clean.csv")
```

```python
# Loading the LEIE exclusion list
d = pd.read_csv("/dsa/groups/casestudycf25/team02/leie_with_valid_npi_clean.csv",dtype={"Suplr_Prvdr_RUCA": "float64"}, na_values=["NA"])
```

```python
#Attribute Names

df.columns
```

```python
# Selecting only Amount based variables
amt = ['Rfrg_NPI','Avg_Suplr_Sbmtd_Chrg', 'Avg_Suplr_Mdcr_Alowd_Amt','Avg_Suplr_Mdcr_Pymt_Amt', 'Avg_Suplr_Mdcr_Stdzd_Amt']
df1 = df[amt]
df1
```

```python
#Attribute Names

df1.columns
```

```python
# Grouping by NPI and computing mean, sum, median, std, min, max
cols_to_agg = ['Avg_Suplr_Sbmtd_Chrg','Avg_Suplr_Mdcr_Alowd_Amt','Avg_Suplr_Mdcr_Pymt_Amt','Avg_Suplr_Mdcr_Stdzd_Amt']

npi_stats = df1.groupby('Rfrg_NPI')[cols_to_agg].agg(['mean', 'sum', 'median', 'std', 'min', 'max'])

npi_stats
```

```python
# Flattening the multindex to make it look pretty
npi_stats.columns = ['_'.join(col).strip() for col in npi_stats.columns.values]
npi_stats = npi_stats.reset_index()
npi_stats.head()
```

```python
# Flag NPIs in the dataset that are listed in the LEIE exclusion list
npi_e = d["npi"]
npi_stats["Excluded"] = npi_stats['Rfrg_NPI'].isin(npi_e).map({True: "y", False: "n"})
npi_stats.head()
```

```python
# Saving file
npi_stats.to_csv("/dsa/groups/casestudycf25/team02/DMEPOS_Amount_stats.csv",index=False)
```
