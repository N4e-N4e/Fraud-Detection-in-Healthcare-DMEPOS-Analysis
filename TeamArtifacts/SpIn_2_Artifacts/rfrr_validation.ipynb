# DMEPOS by Referring Provider validation

We determine that the data fields `Rfrg_NPI` and `Year` comprise a candidate key for the dataset.

```python
import pandas as pd

df = pd.read_csv('/dsa/groups/casestudycf25/team02/DMEPOS_rfrr_clean.csv',dtype={'Rfrg_Prvdr_State_FIPS':str,'Rfrg_Prvdr_Zip5':str}) # ensure Rfrg_Prvdr_State_FIPS & Rfrg_Prvdr_Zip5 are imported as str
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
# check that Rfrg_NPI and Year form a joint key

# get number of rows
nrows_orig = df.shape[0]

df_grouped = df.groupby(["Rfrg_NPI","Year"])[["Tot_Suplr_Benes"]].count()

nrows_grouped = df_grouped.shape[0]

print(f'Original: {nrows_orig}, Grouped: {nrows_grouped}')
display(nrows_orig == nrows_grouped)
```
