# EDA for the DMEPOS by Referring Provider (No Service)

This script uses the DMEPOS_rfrr_clean.csv and leie_with_valid_npi_clean.csv  to create summary stats and scatter plots.

This script creates the "DMEPOS_rfrr_clean_labeled.csv"

```python
import pandas as pd

df = pd.read_csv('/dsa/groups/casestudycf25/team02/DMEPOS_rfrr_clean.csv',dtype={'Rfrg_Prvdr_State_FIPS':str,'Rfrg_Prvdr_Zip5':str}) # ensure Rfrg_Prvdr_State_FIPS & Rfrg_Prvdr_Zip5 are imported as str
df.head()
```

```python
labs = pd.read_csv('/dsa/groups/casestudycf25/team02/leie_with_valid_npi_clean.csv',dtype={'npi':int}) # make sure npi numbers are read as int
labs['EXCLDATE'] = labs.excldate.apply(lambda x: int(x[:4]))
labs = labs[['npi','fraud_flag','EXCLDATE']]
labs.head()
```

```python
# merge labs w/ df on npi, retain all NPIs in df
df = pd.merge(labs, df, how='right', left_on='npi', right_on='Rfrg_NPI')
df.head()
```

```python
# impute missing values
df['npi'] = df['npi'].fillna(df['Rfrg_NPI'].astype(int))
df['fraud_flag'] = df['fraud_flag'].fillna(0)
df['EXCLDATE'] = df['EXCLDATE'].fillna(0)
# change npi to integer type
df['npi'] = df.npi.astype(int)

df.head()
```

```python
def make_labels(fraud_flag, start_exc, data_yr):
    if start_exc >= data_yr:
        return fraud_flag
    else:
        return 0

df['target'] = df[['fraud_flag','EXCLDATE','Year']].apply(lambda x: make_labels(*x), axis=1)
df.head()
```

```python
# convert target to int and drop unnecessary cols
df = df.drop(['fraud_flag','EXCLDATE','Rfrg_NPI'], axis=1)
df['target'] = df.target.astype(int)

# save as csv
df.to_csv('/dsa/groups/casestudycf25/team02/DMEPOS_rfrr_clean_labeled.csv')

df.head()
```

```python
# display summary stats
pd.set_option('display.max_columns', None)
pd.options.display.float_format = '{:,.2f}'.format 

df.iloc[:,17:-2].describe()
```

```python
# summary stats for fraud class
df_fraud = df[df.target==1]

df_fraud.iloc[:,17:-2].describe()
```

Fraudulent providers tend to submit claims for services with higher allowed amounts and receive higher payments compared to the general population of providers. Also, their beneficiary risk scores tend to be higher as well.
